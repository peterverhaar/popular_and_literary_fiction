```python
import os
import json
from os.path import join
import re
from langdetect import detect
```

```python
directory = 'Full_reviews'

one_popular = open(f'1star_popular.txt','w',encoding='utf-8')
two_popular = open(f'2star_popular.txt','w',encoding='utf-8')
three_popular = open(f'3star_popular.txt','w',encoding='utf-8')
four_popular = open(f'4star_popular.txt','w',encoding='utf-8')
five_popular = open(f'5star_popular.txt','w',encoding='utf-8')

one_literary = open(f'1star_literary.txt','w',encoding='utf-8')
two_literary = open(f'2star_literary.txt','w',encoding='utf-8')
three_literary = open(f'3star_literary.txt','w',encoding='utf-8')
four_literary = open(f'4star_literary.txt','w',encoding='utf-8')
five_literary = open(f'5star_literary.txt','w',encoding='utf-8')

json_data = json.load(open('reviews.json',encoding='utf-8'))
for review in json_data:

    if review['type'] == 'literary fiction':
        
        

        if review['rating'] == 1:
            one_literary.write(f'{review["text"]}\n\n')
        elif review['rating'] == 2:
            two_literary.write(f'{review["text"]}\n\n')
        elif review['rating'] == 3:
            three_literary.write(f'{review["text"]}\n\n')
        elif review['rating'] == 4:
            four_literary.write(f'{review["text"]}\n\n')
        elif review['rating'] == 5:
            five_literary.write(f'{review["text"]}\n\n')

    if review['type'] == 'popular fiction':

        if review['rating'] == 1:
            one_popular.write(f'{review["text"]}\n\n')
        elif review['rating'] == 2:
            two_popular.write(f'{review["text"]}\n\n')
        elif review['rating'] == 3:
            three_popular.write(f'{review["text"]}\n\n')
        elif review['rating'] == 4:
            four_popular.write(f'{review["text"]}\n\n')
        elif review['rating'] == 5:
            five_popular.write(f'{review["text"]}\n\n')

one_literary.close()
two_literary.close()
three_literary.close()
four_literary.close()
five_literary.close()
            
one_popular.close()
two_popular.close()
three_popular.close()
four_popular.close()
five_popular.close()
        
```
