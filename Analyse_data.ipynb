```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
books_data = pd.read_csv('goodreads_data.csv')
books_data.columns.tolist()
```

```python
pf = books_data.query( "type=='popular fiction'").shape[0]
print(f"Literary fiction: {pf}")
lf = books_data.query( "type=='literary fiction'").shape[0]
print(f"Popular ficton: {lf}")
```

```python
x_axis = 'text_reviews_count'
y_axis = 'ratings_count'
size = 'average_rating'

fig = plt.figure( figsize = ( 15,15 ))


## This line adds spacing in between the lines of the legend 
sns.set(rc = {'legend.labelspacing': 1.6})

ax = sns.scatterplot( data=books_data, 
                     x=x_axis, 
                     y=y_axis,  
                     hue= 'type',  
                     s = 180 ,
                     palette = ['#641f9c','#f5d442'],
                     alpha = 0.6 )


ax.set_xlabel( 'Number of text reviews'  , fontsize = 16 )
ax.set_ylabel( 'Number of Ratings'  , fontsize = 16 )
ax.set_title( 'Engagement on Goodreads' , fontsize=24 )

for index, row in books_data.iterrows():
    if row[x_axis] > 200000 and row[y_axis] > 2000000:
        plt.text( row[x_axis],row[y_axis],row['title'], 
                 fontsize=12.8,  
                  weight='regular')

# this next line places the legend outside of the graph
plt.legend(bbox_to_anchor=(1.05, 1), loc=2, borderaxespad=0.);

plt.savefig( 'scatterplot.png' , dpi=300 )
```

```python
fig = plt.figure( figsize = ( 8, 5))

graph = sns.boxplot(data=books_data, x = 'type' , y = 'average_rating' , palette = 'crest' )

graph.set_xlabel('')
graph.set_ylabel('Rating')

plt.savefig( 'rating.png' , dpi=300 )
```
